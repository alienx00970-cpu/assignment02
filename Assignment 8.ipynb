```python
# Question # 01
# Make a class called Employee. The init() method for Employee should store its different attributes: name, number, address, salary, post, year of 
# joining, education, and his year of experience.
# Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method 
# called increase_salary() that do increamnet in it's salary.
# Make an instance called employee101 from your class. Print the four attri- butes individually, and then call all it's methods for describing employee,
# salary increamnet and increase year of experience.
# Create 10 different instances from the class, and call describe_employee() for each instance. Then do some updations with it's methods.
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        """Initialize employee attributes"""
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience
    
    def describe_employee(self):
        """Print detailed information about the employee"""
        print("\n" + "="*50)
        print(f"Employee Details:")
        print(f"Name: {self.name}")
        print(f"Employee Number: {self.number}")
        print(f"Address: {self.address}")
        print(f"Salary: ${self.salary:,.2f}")
        print(f"Post/Position: {self.post}")
        print(f"Year of Joining: {self.year_of_joining}")
        print(f"Education: {self.education}")
        print(f"Years of Experience: {self.year_of_experience}")
        print("="*50)
    
    def update_year_of_experience(self, new_experience):
        """Update the years of experience"""
        self.year_of_experience = new_experience
        print(f"Updated experience for {self.name}: {self.year_of_experience} years")
    
    def increase_salary(self, increment_amount):
        """Increase salary by the specified amount"""
        self.salary += increment_amount
        print(f"Salary increased for {self.name}. New salary: ${self.salary:,.2f}")

# Create employee101 instance
employee101 = Employee(
    "John Smith", 
    "EMP001", 
    "123 Main St, New York", 
    75000, 
    "Software Engineer", 
    2020, 
    "B.S. Computer Science", 
    4
)

# Print four attributes individually
print("\n--- Individual Attributes of employee101 ---")
print(f"Name: {employee101.name}")
print(f"Employee Number: {employee101.number}")
print(f"Address: {employee101.address}")
print(f"Salary: ${employee101.salary:,.2f}")

# Call all methods
employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience(5)

# Create 10 different instances
print("\n" + "="*50)
print("CREATING 10 EMPLOYEE INSTANCES")
print("="*50)

employees = [
    Employee("Alice Johnson", "EMP002", "456 Oak Ave, LA", 68000, "Data Analyst", 2021, "M.S. Statistics", 3),
    Employee("Bob Williams", "EMP003", "789 Pine Rd, Chicago", 82000, "Senior Developer", 2018, "M.S. Computer Science", 6),
    Employee("Carol Davis", "EMP004", "321 Elm St, Houston", 59000, "Junior Developer", 2022, "B.S. IT", 2),
    Employee("David Brown", "EMP005", "654 Maple Dr, Phoenix", 95000, "Team Lead", 2015, "MBA", 9),
    Employee("Eva Martinez", "EMP006", "987 Cedar Ln, Philadelphia", 71000, "DevOps Engineer", 2019, "B.S. Engineering", 5),
    Employee("Frank Wilson", "EMP007", "147 Birch Blvd, San Antonio", 63000, "QA Engineer", 2020, "B.S. QA", 4),
    Employee("Grace Lee", "EMP008", "258 Spruce Way, San Diego", 88000, "Product Manager", 2017, "MBA", 7),
    Employee("Henry Taylor", "EMP009", "369 Ash Ct, Dallas", 54000, "Support Engineer", 2023, "B.S. IT", 1),
    Employee("Ivy Chen", "EMP010", "741 Willow Pkwy, San Jose", 92000, "Architect", 2016, "Ph.D. CS", 8),
    Employee("Jack Robinson", "EMP011", "852 Poplar St, Austin", 67000, "Frontend Developer", 2021, "B.S. Design", 3)
]

# Call describe_employee for each instance
for emp in employees:
    emp.describe_employee()

# Perform some updates
print("\n" + "="*50)
print("PERFORMING UPDATIONS")
print("="*50)

employees[0].increase_salary(3000)
employees[0].update_year_of_experience(4)

employees[2].increase_salary(2000)
employees[2].update_year_of_experience(3)

employees[4].increase_salary(5000)
employees[4].update_year_of_experience(6)

employees[7].increase_salary(1500)
employees[7].update_year_of_experience(2)

# Display updated info for the modified employees
print("\n--- Updated Information ---")
for emp in [employees[0], employees[2], employees[4], employees[7]]:
    emp.describe_employee()
```
