### Question # 02
### Currency Converter

```python
# Question # 02
# Create a Python function called currency_converter() that takes two arguments:
# amount: The amount of money to be converted (in USD).
# exchange_rate (default is 280): The conversion rate from USD to another currency (e.g., PKR). You can assume the default rate is 1 USD = 280 PKR.
# The function should return a message showing how much the given amount is worth in the target currency (PKR).
def currency_converter(amount, exchange_rate=280):
    """Convert USD to PKR"""
    converted_amount = amount * exchange_rate
    return f"{amount} USD is equivalent to {converted_amount} PKR"

# Test the function
print(currency_converter(100))
print(currency_converter(50, 300))
```

### Question # 03
### Display Student info

```python
# Question # 03 
#  Write a Python function display_student_info that accepts the following details about a student:
# name (Keyword argument): The name of the student.
# age (Keyword argument): The age of the student.
# grade (Keyword argument): The grade the student is in.
# school (Keyword argument): The name of the school the student attends.
def display_student_info(**kwargs):
    """Display student information using keyword arguments"""
    print(f"Name: {kwargs.get('name', 'Not provided')}")
    print(f"Age: {kwargs.get('age', 'Not provided')}")
    print(f"Grade: {kwargs.get('grade', 'Not provided')}")
    print(f"School: {kwargs.get('school', 'Not provided')}")

# Test the function
display_student_info(name="John Doe", age=15, grade="10th", school="Greenwood High")
```

### Question # 05
###  Calculate rectangle with nested functions

```python
# Question # 05
# Write a Python function calculate_rectangle that accepts the length and width of a rectangle as arguments. Inside this function, create two nested
# functions:
# calculate_area: This function should calculate the area of the rectangle.
# calculate_perimeter: This function should calculate the perimeter of the rectangle.
# The main function should return both the area and the perimeter of the rectangle.
def calculate_rectangle(length, width):
    """Calculate area and perimeter using nested functions"""
    
    def calculate_area():
        return length * width
    
    def calculate_perimeter():
        return 2 * (length + width)
    
    area = calculate_area()
    perimeter = calculate_perimeter()
    
    return area, perimeter

area, perimeter = calculate_rectangle(5, 3)
print(f"Area: {area}")
print(f"Perimeter: {perimeter}")
```

### Question # 06
###  List Comprehension numbers divisible by 3 or 5

```python
# Question # 06
# Create a list of numbers, generate a new list containing only the numbers that are divisible by 3 or 5. (List Comprehension)
numbers = list(range(1, 51))

divisible_by_3_or_5 = [num for num in numbers if num % 3 == 0 or num % 5 == 0]

print("Numbers divisible by 3 or 5:")
print(divisible_by_3_or_5)
```

### Question # 07
### ATM Withdrawal Program



```python
# Question # 07
# Write a Python program that:
# Has a balance of 1000.
# Asks the user for a withdrawal amount.
# Checks for these errors:
# Non-numeric input → print "Please enter a valid number."
# Amount ≤ 0 → print "Amount must be more than zero."
# Amount > balance → print "Insufficient balance."
# If valid, deducts amountbalance = 1000
balance = 1000

try:
    amount = input("Enter amount to withdraw: ")
    
    if not amount.replace('.', '').isdigit():
        print("Please enter a valid number.")
    else:
        amount = float(amount)
        if amount <= 0:
            print("Amount must be more than zero.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            print(f"Withdrawal successful. Remaining balance: {balance}")
            
except ValueError:
    print("Please enter a valid number.")
finally:
    print("Thank you for using the ATM.")
```

### Question # 08
### Ticket Booking Program

```python
# Question # 08
# Write a Python program that:
# Has 5 seats available (available_seats = 5).
# Asks the user how many tickets they want to book.
# Checks for these cases:
# Non-numeric input → print "Please enter a valid number."
# Tickets ≤ 0 → print "You must book at least 1 ticket."
# Tickets more than available seats → print "Not enough seats available."
# If valid, confirms the booking by printing how many tickets were booked.
# Always prints "Thank you for booking with us." at the end.
# Ticket Booking Program
available_seats = 5

tickets = input("How many tickets do you want to book? ")

if not tickets.isdigit():
    print("Please enter a valid number.")
else:
    tickets = int(tickets)
    if tickets <= 0:
        print("You must book at least 1 ticket.")
    elif tickets > available_seats:
        print("Not enough seats available.")
    else:
        available_seats -= tickets
        print(f"{tickets} ticket(s) booked successfully.")

print("Thank you for booking with us.")
```

### Question # 10
### Student Class

```python
# Question # 10
# Define a class named Student that has the following:
# Attributes: name, grade, and age.
# A method introduce() that prints:
# "My name is {name}, I am {age} years old."
# A method show_grade() that prints:
# "My grade is {grade}."
# Create an object of the Student class with sample data (e.g., "Alice", "A", 16), and call both methods on the object.
class Student:
    def __init__(self, name, grade, age):
        self.name = name
        self.grade = grade
        self.age = age
    
    def introduce(self):
        print(f"My name is {self.name}, I am {self.age} years old.")
    
    def show_grade(self):
        print(f"My grade is {self.grade}.")

student1 = Student("Alice", "A", 16)
student1.introduce()
student1.show_grade()
```
